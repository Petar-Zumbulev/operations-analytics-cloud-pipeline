```python
# Calculate the total quantity ordered across all rows in order_items

# Berechnen Sie die gesamte bestellte Menge über alle Zeilen der Tabelle order_items

query = """

SELECT
    SUM(quantity) AS total_quantity
FROM order_items;

"""

df = pd.read_sql_query(query, conn)
print(df)

```

I use SUM(quantity) to add the ordered quantity across all order-item rows. I assign the result the alias total_quantity so the output column is clear and readable. Since I want one overall total, no GROUP BY is needed

Ich verwende SUM(quantity), um die bestellte Menge über alle Bestellpositionen zu addieren. Mit dem Alias total_quantity erhält die Ergebnisspalte einen klaren und verständlichen Namen. Da ich nur einen Gesamtwert berechnen möchte, brauche ich kein GROUP BY

```python
# Calculate the total quantity ordered for each product

# Berechnen Sie die gesamte bestellte Menge für jedes Produkt

query = """

SELECT
    product_id,
    SUM(quantity) AS total_quantity
FROM order_items
GROUP BY product_id
ORDER BY total_quantity DESC;

"""

df = pd.read_sql_query(query, conn)
print(df)

```

I group the rows by product_id because I want one result per product. Then I sum the quantity within each product group. Finally, I sort by the calculated total in descending order so the product with the highest quantity appears first

Ich gruppiere die Zeilen nach product_id, weil ich ein Ergebnis pro Produkt erhalten möchte. Anschließend summiere ich die Menge innerhalb jeder Produktgruppe. Zum Schluss sortiere ich die berechnete Gesamtmenge absteigend, sodass das Produkt mit der höchsten Menge zuerst erscheint

```python
# Calculate the total revenue for each product

# Berechnen Sie den Gesamtumsatz für jedes Produkt

query = """

SELECT
    product_id,
    SUM(quantity * unit_price) AS Gesamtumsatz
FROM order_items
GROUP BY product_id
ORDER BY  Gesamtumsatz DESC;

"""

df = pd.read_sql_query(query, conn)
print(df)

```

I first calculate the revenue for each order-item row by multiplying quantity by unit_price. Then I sum those values for each product using GROUP BY product_id. Finally, I sort the total revenue in descending order so the highest-revenue product appears first

Zuerst berechne ich den Umsatz für jede Bestellposition, indem ich quantity mit unit_price multipliziere. Anschließend summiere ich diese Werte für jedes Produkt mit GROUP BY product_id. Zum Schluss sortiere ich den Gesamtumsatz absteigend, sodass das umsatzstärkste Produkt zuerst erscheint

```python
# Calculate the total quantity ordered for each product, 
# but return only products with a total quantity greater than 200

# Berechnen Sie die gesamte bestellte Menge für jedes Produkt, 
# aber geben Sie nur Produkte zurück, deren Gesamtmenge größer als 200 ist

query = """

SELECT
    product_id,
    SUM(quantity) AS total_quantity
FROM order_items
GROUP BY product_id
HAVING total_quantity > 200;

"""

df = pd.read_sql_query(query, conn)
print(df)
```

The key difference

WHERE filters individual rows before grouping.

HAVING filters groups after aggregation.

I group the rows by product_id and calculate the total quantity for each product using SUM(quantity). Because I want to filter the aggregated product totals rather than the original rows, I use HAVING instead of WHERE. Finally, I sort the remaining products by total quantity in descending order

Ich gruppiere die Zeilen nach product_id und berechne mit SUM(quantity) die Gesamtmenge für jedes Produkt. Da ich die aggregierten Produktsummen und nicht die ursprünglichen Einzelzeilen filtern möchte, verwende ich HAVING statt WHERE. Zum Schluss sortiere ich die verbleibenden Produkte nach der Gesamtmenge absteigend

```python
# Show every order together with the corresponding customer name

# Zeigen Sie jede Bestellung zusammen mit dem dazugehörigen Kundennamen

query = """

SELECT
     orders.order_id,
     orders.order_date,
     orders.status,
     customers.customer_name
FROM orders
JOIN customers
    ON orders.customer_id = customers.customer_id
ORDER BY orders.order_id ASC;

"""

df = pd.read_sql_query(query, conn)
print(df)
```

I join the orders table with the customers table using customer_id, because that column connects each order to its customer. Then I select the order information and the corresponding customer name. Finally, I sort the result by order_id in ascending order

A normal JOIN here means INNER JOIN.

That means only rows with a matching customer_id in both tables are returned.

Ich verbinde die Tabelle orders mit der Tabelle customers über customer_id, weil diese Spalte jede Bestellung dem entsprechenden Kunden zuordnet. Danach wähle ich die Bestelldaten und den zugehörigen Kundennamen aus. Zum Schluss sortiere ich das Ergebnis nach order_id aufsteigend

```python
# Calculate the total revenue for each order

# Berechnen Sie den Gesamtumsatz für jede Bestellung

query = """

SELECT
     order_id,
     SUM(quantity * unit_price) AS total_revenue
FROM order_items
GROUP BY order_id
ORDER BY total_revenue DESC;

"""

df = pd.read_sql_query(query, conn)
print(df)
```

```python
# Calculate the total revenue for each customer

query = """

SELECT
     customers.customer_name,
     SUM(order_items.quantity * order_items.unit_price) AS total_revenue
FROM order_items
JOIN orders
    ON order_items.order_id = orders.order_id
JOIN customers
    ON orders.customer_id = customers.customer_id
GROUP BY customers.customer_name
ORDER BY total_revenue DESC;

"""

# the grain in the final table is one row per customer

df = pd.read_sql_query(query, conn)
print(df)
```

```python
# Calculate the total revenue for each customer, for delivered orders

query = """

SELECT
     customers.customer_name,
     SUM(order_items.quantity * order_items.unit_price) AS total_revenue
FROM order_items
JOIN orders
    ON order_items.order_id = orders.order_id
JOIN customers
    ON orders.customer_id = customers.customer_id
WHERE orders.status = 'delivered'
GROUP BY customers.customer_name
ORDER BY total_revenue DESC;

"""

# the grain in the final table is one row per customer

# HAVING is used to filter aggregations
# But status is a normal row-level column, not an aggregated value
# so this should be filtered with WHERE -before- the grouping

# SQL filters out non-delivered orders before calculating the revenue per customer

df = pd.read_sql_query(query, conn)
print(df)
```

```python
# Find all orders that do not have a corresponding shipment

query = """

SELECT
     orders.order_id,
     orders.order_date,
     orders.status,
     shipments.shipment_id,
     shipments.shipped_date
FROM orders
LEFT JOIN shipments
    ON orders.order_id = shipments.order_id
WHERE shipments.shipment_id IS NULL
ORDER BY orders.order_id ASC;

"""

df = pd.read_sql_query(query, conn)
print(df)
```

```python
# Find all shipments where delivered_date is missing

query = """

SELECT
     shipment_id,
     order_id,
     shipped_date,
     delivered_date
FROM shipments
WHERE delivered_date IS NULL
ORDER BY shipment_id ASC;

"""

df = pd.read_sql_query(query, conn)
print(df)
```

```python
# Check whether any order_id appears more than once in the orders table.

query = """

SELECT
     order_id,
     COUNT(*) AS duplicate_count
FROM orders
GROUP BY order_id
HAVING COUNT(*) > 1;
"""

df = pd.read_sql_query(query, conn)
print(df)
```

I would first identify duplicates using GROUP BY and HAVING COUNT(*) > 1. If duplicates exist, I would investigate whether they're expected business duplicates or data-quality issues before deciding how to remove or consolidate them.

```python
# Create a new column order_status which describes the status

query = """

SELECT
    order_id,
    status,

    CASE
        WHEN status = 'delivered'
            THEN 'Completed'

        WHEN status = 'open'
            THEN 'In Progress'

        ELSE 'Cancelled'
    END AS order_status

FROM orders
ORDER BY order_id;
"""

df = pd.read_sql_query(query, conn)
print(df)
```
