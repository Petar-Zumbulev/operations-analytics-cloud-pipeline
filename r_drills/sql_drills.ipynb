```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("hassia_practice.db")
```

```python
query = """
SELECT *
FROM orders;
"""

df = pd.read_sql_query(query, conn)
print(df)
```

```python
query = """

SELECT 
    order_id,
    customer_id,
    order_date,
    status
FROM orders
ORDER BY order_date ASC;

"""

df = pd.read_sql_query(query, conn)
print(df)
```

```python
query = """

SELECT 
    order_id,
    customer_id,
    order_date,
    status
FROM orders
ORDER BY order_date DESC;

"""

df = pd.read_sql_query(query, conn)
print(df)
```

```python
query = """

SELECT 
    COUNT(*) AS number_of_orders
FROM orders;

"""

# * means count every row 

df = pd.read_sql_query(query, conn)
print(df)
```

```python
query = """

SELECT 
    *
FROM orders;

"""

df = pd.read_sql_query(query, conn)
print(df)
```

```python
query = """

SELECT 
    COUNT(*) AS delivered_orders
FROM orders
WHERE status = 'delivered';

"""

df = pd.read_sql_query(query, conn)
print(df)
```

```python
# REALLY USEFUL

query = """

SELECT 
    status,
    COUNT(*) AS number_of_orders
FROM orders
GROUP BY status;

"""

# split the table by status and count rows inside each status group

df = pd.read_sql_query(query, conn)
print(df)
```

```python
query = """

SELECT
    status,
    COUNT(*) AS number_of_orders
FROM orders
GROUP BY status
ORDER BY number_of_orders DESC;

"""

# whatever comes after SELECT is what columns show up in the result

df = pd.read_sql_query(query, conn)
print(df)
```

```python
query = """

SELECT
    SUM(quantity) AS total_quantity
FROM order_items;

"""

df = pd.read_sql_query(query, conn)
print(df)
```

COUNT(*) counts rows

SUM(column) adds numbers inside the column

```python
query = """

SELECT
    SUM(quantity * unit_price) AS total_revenue
FROM order_items;

"""

# As long as I enter the column names within the SUM() function, 
# sql knows to do the operation row by row. This is the case in general

# SUM(quantity * unit_price) means For each row, calculate 
# quantity * unit_price, then add all those results together

df = pd.read_sql_query(query, conn)
print(df)
```

```python
query = """

SELECT
    order_id,
    SUM(quantity * unit_price) AS revenue
FROM order_items
GROUP BY order_id;

"""

df = pd.read_sql_query(query, conn)
print(df)
```

Important pattern:

SELECT
    group_column,
    SUM(number_column_or_calculation) AS new_name
FROM table
GROUP BY group_column;

```python
# Show each order together with the customer name

query = """

SELECT
    orders.order_id,
    customers.customer_name,
    orders.order_date,
    orders.status
FROM orders
JOIN customers
    ON orders.customer_id = customers.customer_id;

"""

# Ich verbinde die Tabelle orders mit der Tabelle customers 
# über die gemeinsame Spalte customer_id. Dadurch bekomme ich 
# zu jeder Bestellung den passenden Kundennamen. Ich nutze 
# dafür einen JOIN mit einer ON-Bedingung, damit nicht jede 
# Bestellung mit jedem Kunden kombiniert wird.

df = pd.read_sql_query(query, conn)
print(df)
```

```python
# Show all delivered orders together with the customer name

query = """

SELECT
    orders.order_id,
    customers.customer_name,
    orders.order_date,
    orders.status
FROM orders
JOIN customers
    ON orders.customer_id = customers.customer_id
WHERE status = 'delivered';

"""

df = pd.read_sql_query(query, conn)
print(df)
```

```python
# Calculate the revenue per order, but only for delivered orders

query = """

SELECT
    orders.order_id,
    orders.status,
    SUM(order_items.quantity * order_items.unit_price) AS revenue
FROM orders
JOIN order_items
    ON orders.order_id = order_items.order_id
WHERE orders.status = 'delivered'
GROUP BY orders.order_id, orders.status;

"""

# General SQL rule:

# Every normal column in SELECT should either be inside GROUP BY or inside an aggregate function like SUM(), COUNT(), AVG().

df = pd.read_sql_query(query, conn)
print(df)
```
